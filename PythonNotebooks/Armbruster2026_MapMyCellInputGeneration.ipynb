### -------- Armbruster2026_MapMyCellInputGeneration--------

Version: 2026-07-30


- In this script we loaded the raw count matrix of our annotated zebrafish CA clusters that we exported in "Script8_Armbruster2026_Preparation_for_MapMyCell" and change the zebrafish gene names into their respective mouse ortholog names based on orthology information from https://www.alliancegenome.org/downloads#orthology downladed in March 2025 (see Table S5A).

- Raw counts of zebrafish paralogous genes with a single suggested mouse homolog were summed up. 

- Then the renamed count matrix containing 1,922 cells and 11,490 genes was exported as csv file and forwarded to the browser-based version of MapMyCells (Allen Institute; RRID:SCR_024672).

- The session info including the used packages and their versions is provided at the end of this script.

# 3. Load data
- Now we replace the zebrafish gene names with their mouse homologs.
- Duplicated zebrafish genes, will be summed up (e.g. pax6a+pax6b = Pax6)

```python
homology_zf_mouse_df[0:10]
```

```python
#zf vs mouse
homology_zf_mouse_dict = dict(zip(homology_zf_mouse_df["zf10_gene"], homology_zf_mouse_df["mouse_gene"]))
```

- Filter for genes that exist in the homology dict (i.e., have a match)

```python
zf_mouse_genes_with_homologs = expr_df.index.intersection(homology_zf_mouse_dict.keys())
expr_df_zf_mouse_filtered = expr_df.loc[zf_mouse_genes_with_homologs]
```

- Rename zebrafish genes to mouse homologs and sum up the expression values of duplicates

```python
expr_df_zf_mouse_renamed = expr_df_zf_mouse_filtered.rename(index=homology_zf_mouse_dict)
expr_df_zf_mouse_renamed = expr_df_zf_mouse_renamed.groupby(expr_df_zf_mouse_renamed.index).sum()
```

- Check renamed expression matrix

```python
expr_df_zf_mouse_renamed[0:10]
```

# 4. Save renamed gene expression matrix

```python
expr_df_zf_mouse_renamed.to_csv(r"D:\scRNAseq\10X\DA136\SeuratAnalysis_2024-09-03\MapMyCells\MapMyCells_input_data\modified_expression_matrix_mousenames.csv")
```
